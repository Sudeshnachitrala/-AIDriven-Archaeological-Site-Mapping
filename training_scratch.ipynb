```python
import yaml, os, glob

# ✅ Update this line to your correct dataset folder
dataset_path = "/content/extracted_files"   # or whatever you found above

yaml_path = os.path.join(dataset_path, "data.yaml")
os.makedirs(dataset_path, exist_ok=True)   # ensure the folder exists

if not os.path.exists(yaml_path):
    label_files = glob.glob(os.path.join(dataset_path, "train/labels", "*.txt"))
    class_ids = set()
    for lf in label_files:
        with open(lf, 'r') as f:
            for line in f:
                parts = line.strip().split()
                if parts:
                    class_ids.add(int(parts[0]))
    nc = max(class_ids) + 1 if class_ids else 1
    names = [f"class{i}" for i in range(nc)]
    yaml_content = {
        'path': dataset_path,
        'train': 'train/images',
        'val': 'valid/images',
        'test': 'test/images',
        'nc': nc,
        'names': names
    }
    with open(yaml_path, 'w') as f:
        yaml.dump(yaml_content, f)
    print("✅ Created data.yaml:", yaml_content)
else:
    with open(yaml_path, 'r') as f:
        print("Existing data.yaml:\n", f.read())
```

```python
# 1️⃣ Set WandB offline BEFORE importing YOLO
import os
os.environ["WANDB_MODE"] = "offline"

# 2️⃣ Install / upgrade YOLOv8
!pip install -U ultralytics

# 3️⃣ Import YOLO AFTER setting offline mode
from ultralytics import YOLO

# 4️⃣ Initialize YOLO model (architecture only)
model = YOLO("yolov8n.yaml")

# 5️⃣ Path to your dataset
data_yaml = "/content/extracted_files/data.yaml"

# 6️⃣ Train from scratch
model.train(
    data=data_yaml,
    epochs=50,
    batch=16,
    imgsz=640,
    device=0,
    pretrained=False,
    project="yolo_training",  # valid project name
    name="vegetation_scratch",
    exist_ok=True,
    save_dir="/content/drive/MyDrive/yolo_training/vegetation_scratch"  # save outputs
)
```
